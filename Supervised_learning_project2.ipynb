```python
for col in num_cols:
    Q1 = life_data[col].quantile(0.25)
    Q3 = life_data[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((life_data[col] < lower) | (life_data[col] > upper)).sum()

    print(col, ":", outliers)
```

```python
 #Drop target column from num_column
num_cols = life_data.drop('Life expectancy ', axis=1).select_dtypes(include='number').columns

# IQR method + Capping
for col in num_cols:
    Q1 = life_data[col].quantile(0.25)
    Q3 = life_data[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    life_data[col] = life_data[col].clip(lower=lower, upper=upper)
```

# Train Test Split

```python
y = life_data['Life expectancy '].reset_index(drop=True)#dependent feature or target column
x = life_data.drop('Life expectancy ',axis = 1).reset_index(drop=True)
```

```python
#train test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state = 42)
```

# Scaling

```python
# non scaled x_train and x_test
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Linear regression with scaled data
lr_model_scaled = LinearRegression()
lr_model_scaled.fit(x_train1, y_train)
y_pred_lr_model_scaled = lr_model_scaled.predict(x_test1)

# Regression evaluation metrics
print('R2 Score is: ', r2_score(y_test, y_pred_lr_model_scaled))
print('MAE is: ', mean_absolute_error(y_test, y_pred_lr_model_scaled))
print('RMSE is: ', mean_squared_error(y_test, y_pred_lr_model_scaled) ** 0.5)
```

# Decision Tree

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Initialize and train Decision Tree Regressor
dt_model = DecisionTreeRegressor(random_state=42)
dt_model.fit(x_train, y_train)

# Make predictions on test set
y_pred_dt = dt_model.predict(x_test)

# Evaluation metrics
print('Decision Tree Metrics:')
print('R2 Score is: ', r2_score(y_test, y_pred_dt))
print('MAE is: ', mean_absolute_error(y_test, y_pred_dt))
print('RMSE is: ', mean_squared_error(y_test, y_pred_dt) ** 0.5)
```

# Random Forest

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Initialize and train Random Forest Regressor
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model.fit(x_train, y_train)

# Predictions on test set
y_pred_rf = rf_model.predict(x_test)

# Evaluation metrics
print('Random Forest Metrics:')
print('R2 Score is: ', r2_score(y_test, y_pred_rf))
print('MAE is: ', mean_absolute_error(y_test, y_pred_rf))
print('RMSE is: ', mean_squared_error(y_test, y_pred_rf) ** 0.5)
```

Random forest has highest accuracy
